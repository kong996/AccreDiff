# 01 - Constants and Setup

This notebook introduces the built-in constants and data structures in `accrediff`.  
It covers:
- Element atomic masses (`Elements`)
- Meteorite bulk compositions (`EF_bulk`, `EC_bulk`, `OC_bulk`, `CI_bulk`)
- Planet mantle/core compositions (`Earth_mantle`, `Earth_core`, `Mars_mantle`, `Mars_core`)
- `MolarMassCalculator` basic usage
- Bulk compositions of Earth & Mars

> **Run this notebook first** before proceeding to the rest of the quickstart series.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sys
sys.path.append('../../src')
import accrediff as ad

print(f"accrediff version: {ad.__version__}")
```

## 1. Plot Style Settings

```python
plt.rcParams.update({
    "font.family": "serif",
    "font.size": 12,
    "axes.linewidth": 1,
    "axes.labelsize": 14,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": False,
    "ytick.right": False,
    "figure.dpi": 150
})

Colors = {
    'Earth': "#1f77b4",
    'Mars' : "#d62728",
    'EF'   : "#8c564b",
    'EC'   : "#ff7f0e",
    'OC'   : "#2ca02c",
    'CI'   : "#9467bd",
}
print("Plot style configured.")
print("Color palette:", Colors)
```

## 2. Elements — Atomic Mass Table

```python
Elements = ad.constants.Elements
print(f"Total elements loaded: {len(Elements)}")

# Display as DataFrame
df_elements = pd.DataFrame.from_dict(Elements, orient='index', columns=['Atomic Mass (g/mol)'])
df_elements.index.name = 'Element/Compound'
df_elements
```

## 3. MolarMassCalculator

```python
calc = ad.MolarMassCalculator(mass_table=Elements)

# Calculate molar masses for common oxide/metal compounds
compounds = ['MgO', 'Al2O3', 'CaO', 'FeO', 'NiO', 'SiO2', 'Fe', 'Ni', 'Si', 'O']
for c in compounds:
    Elements[c] = calc.molar_mass(c)
    print(f"  {c:8s}: {Elements[c]:.4f} g/mol")
```

## 4. Meteorite Bulk Compositions

```python
EF_bulk = ad.constants.EF_bulk
EC_bulk = ad.constants.EC_bulk
OC_bulk = ad.constants.OC_bulk
CI_bulk = ad.constants.CI_bulk

df_met = pd.DataFrame({
    'EF': EF_bulk,
    'EC': EC_bulk,
    'OC': OC_bulk,
    'CI': CI_bulk,
})
print("Meteorite bulk compositions (wt%):")
df_met
```

## 5. Planet Mantle & Core Compositions

```python
Earth_mantle = ad.constants.Earth_mantle
Earth_core   = ad.constants.Earth_core
Mars_mantle  = ad.constants.Mars_mantle
Mars_core    = ad.constants.Mars_core

df_planet = pd.DataFrame({
    'Earth_mantle': Earth_mantle,
    'Earth_core'  : Earth_core,
    'Mars_mantle' : Mars_mantle,
    'Mars_core'   : Mars_core,
})
print("Planet mantle & core compositions (wt%):")
df_planet
```

## 6. Bulk compositions of Earth & Mars

```python
Earth_bulk, Mars_bulk = {}, {}
# Calculate bulk compositions for Earth and Mars
mantle_items = ['MgO', 'Al2O3','CaO', 'FeO', 'NiO', 'SiO2']
core_items = ['Fe', 'Ni','Si', 'O']

for item in mantle_items:
    Earth_bulk[item] = Earth_mantle[item] * Earth_mantle['mantle']
    Mars_bulk[item] = Mars_mantle[item] * Mars_mantle['mantle']

for item in core_items:
    Earth_bulk[item] = Earth_core[item] * Earth_core['core']
    Mars_bulk[item] = Mars_core[item] * Mars_core['core']
print("Calculated bulk compositions (wt%):")
df_bulk = pd.DataFrame({
    'Earth_bulk': Earth_bulk,
    'Mars_bulk' : Mars_bulk,
})
df_bulk
```
